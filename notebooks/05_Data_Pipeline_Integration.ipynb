```python
def validate_dataframe(df, name):
    """
    Perform basic quality checks on a DataFrame.
    """

    report = {
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Missing_Values": int(df.isnull().sum().sum()),
        "Duplicate_Rows": int(df.duplicated().sum()),
        "Memory_MB": round(
            df.memory_usage(deep=True).sum() / (1024 ** 2), 3
        )
    }

    return report


# Validate all loaded datasets
validation_results = []

for name, df in project_data.items():

    result = validate_dataframe(df, name)
    validation_results.append(result)


validation_df = pd.DataFrame(validation_results)

display(validation_df)
```

```python
# Verify company-specific returns in abnormal_returns

company_return_columns = ["AAPL", "MSFT", "TSLA"]

print("COMPANY RETURN VALIDATION")
print("-" * 55)

for company in company_return_columns:

    company_data = abnormal_df[
        abnormal_df["company"] == company
    ]

    other_companies = [
        c for c in company_return_columns if c != company
    ]

    active_missing = company_data[company].isna().sum()

    inactive_missing = (
        company_data[other_companies].isna().sum().sum()
    )

    print(f"\nCompany: {company}")
    print("Events:", company_data["event_id"].nunique())
    print("Observations:", len(company_data))
    print("Missing active returns:", active_missing)
    print("Missing inactive returns:", inactive_missing)

    assert active_missing == 0
    assert inactive_missing == len(company_data) * 2

print("\nAll structural missing-value checks passed!")
```

```python
def filter_events(
    df,
    company=None,
    sentiment=None,
    car_direction=None,
    event_id=None
):
    """
    Filter integrated event data using multiple conditions.
    """

    filtered_df = df.copy()

    if company is not None:
        filtered_df = filtered_df[
            filtered_df["company"].isin(
                company if isinstance(company, list) else [company]
            )
        ]

    if sentiment is not None:
        filtered_df = filtered_df[
            filtered_df["FinBERT_Sentiment"].isin(
                sentiment if isinstance(sentiment, list) else [sentiment]
            )
        ]

    if car_direction is not None:
        filtered_df = filtered_df[
            filtered_df["CAR_Direction"].isin(
                car_direction if isinstance(car_direction, list)
                else [car_direction]
            )
        ]

    if event_id is not None:
        filtered_df = filtered_df[
            filtered_df["event_id"].isin(
                event_id if isinstance(event_id, list) else [event_id]
            )
        ]

    return filtered_df.reset_index(drop=True)
```

```python
integrated_df = project_data["integrated_event_data"]

# Example 1: Apple events
apple_events = filter_events(
    integrated_df,
    company="AAPL"
)

# Example 2: Positive sentiment
positive_events = filter_events(
    integrated_df,
    sentiment="positive"
)

# Example 3: Negative CAR direction
negative_car_events = filter_events(
    integrated_df,
    car_direction="negative"
)

# Example 4: Combined filters
combined_events = filter_events(
    integrated_df,
    company="MSFT",
    sentiment="positive"
)

print("FILTER FUNCTION TEST RESULTS")
print("-" * 50)

print("Apple events:", len(apple_events))
print("Positive sentiment events:", len(positive_events))
print("Negative CAR events:", len(negative_car_events))
print("Microsoft positive sentiment events:", len(combined_events))

display(combined_events)
```
